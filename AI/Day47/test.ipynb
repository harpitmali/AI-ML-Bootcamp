# Layer Calculation

```python
import numpy as np 

X = np.array([2.0, 3.0])

W = np.array([
    [0.5, 0.2],
    [0.1, 0.4]
])

b = np.array([1.0, 0.5])

Z = W @ X + b

print("Z: ", Z)
```

## ReLU

```python
A = np.maximum(0, Z)

print("A: ", A)
```

## Sigmoid

```python
A = 1 / (1 + np.exp(Z))

print("A: ", A)
```
