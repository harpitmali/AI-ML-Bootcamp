# Attention — The Core Idea

```python
import numpy as np 
```

```python
X = np.array([
    [1.0, 0.0],
    [0.9, 0.1],
    [0.0, 1.0]
])

Q = X
K = X
V = X

scores = Q @ K.T

print("Scores: ")
print(scores)
```

```python
print(scores.shape)
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)
```

```python
attension_weight = softmax(scores)

print("Attension Weight: ")
print(attension_weight)

print("Rows Sum: ")
print(attension_weight.sum(axis=1))
```

```python
attension_output = attension_weight @ V

print("Attension Output: ")
print(attension_output)

print("Shape: ")
print(attension_output.shape)
```

# Practical Task

```python
X = np.array([
    [1.0, 0.0, 1.0],
    [0.9, 0.1, 0.8],
    [0.0, 1.0, 0.2],
    [0.2, 0.8, 0.1]
])

Q = X
K = X
V = X
```

```python
scores = Q @ K.T

print("Scores: ")
print(scores)

print("Shape: ")
print(scores.shape)
```

```python
attension_weight = softmax(scores)

print("Attension Weight: ")
print(attension_weight)

print("Rows Sum: ")
print(attension_weight.sum(axis=1))
```

```python
output = attension_weight @ V

print("Attension Output: ")
print(output)

print("Shape: ")
print(output.shape)
```
