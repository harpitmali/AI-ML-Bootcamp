# A practical NumPy example

```python
import numpy as np
```

```python
vocal_size = 6
embedding_dim = 4

embedding_matrix = np.random.randn(
    vocal_size,
    embedding_dim
)

print("Embedding Matrix: ")
print(embedding_matrix)
```

```python
token_id = 2

token_vector = embedding_matrix[token_id]

print("Token ID: ", token_id)
print("Token Vector: ", token_vector)
```

## Multiple tokens

```python
token_ids = np.array([2, 4, 1])

vectors = embedding_matrix[token_ids]

print(vectors)
```

## Practical Task

```python
vocab = {
    "I": 0,
    "love": 1,
    "AI": 2,
    "and": 3,
    "Python": 4
}

embedding_dim = 5

embedding_matrix = np.random.randn(
    len(vocab),
    embedding_dim
)
```

```python
text = ["I", "love", "AI"]

token_ids = [vocab[token] for token in text]
print(token_ids)
```

```python
vectors = embedding_matrix[token_ids]
```

```python
print("Token IDs:")
print(token_ids)

print("\nEmbedding vectors:")
print(vectors)

print("\nShape:")
print(vectors.shape)
```

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
```

```python
text = "I love AI"

token_ids = encoding.encode(text)

print("Token IDs: ", token_ids)
print("Number of tokens: ", len(token_ids))
```
