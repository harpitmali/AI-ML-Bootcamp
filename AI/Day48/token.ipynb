# Practical tokenization with tiktoken

```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
```

## Encoding

```python
text = "I am learning AI and machine learning."

tokens = encoding.encode(text)

print(tokens)
```

## Decoding

```python
decoded = encoding.decode(tokens)

print(decoded)
```

## See the individual token pieces

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(token_id, repr(token_text))
```

## Token count

```python
print(f"Total Tokens: {len(tokens)}")
```

## Another Example

```python
texts = [
    "Hello World!",
    "I am learing artificial intelligence.",
    "Transformers are powerful models.",
    "def add(a, b): return a + b"
]

for text in texts:

    token_ids = encoding.encode(text)

    print("\nText: ", text)
    print("Token IDs: ", token_ids)
    print("Token Count: ", len(token_ids))

    print("Tokens: ")
    for token_id in token_ids:
        print(repr(encoding.decode([token_id])), end=" | ")

    print()
```

## one more

```python
text = "orewa monkey d. luffy kaizoku ou ni naru otoko da"

token_ids = encoding.encode(text)

print("Token count:", len(token_ids))

for token_id in token_ids:
    print(token_id, repr(encoding.decode([token_id])))
```

## Mini Experiment

```python
texts = [
    "AI",
    "artificial intelligence",
    "ArtificialIntelligence",
    "machine learning",
    "machinelearning"
]

for text in texts:
    tokens = encoding.encode(text)

    print(
        f"{text!r:30}"
        f"→ {len(tokens)} tokens "
        f"→ {tokens}"
    )
```
