### ML Dataset

```python
import pandas as pd

data = {
    "Experience": [1, 2, 3, 4, 5, 6, 7, 8],
    "Age": [22, 24, 26, 28, 30, 32, 34, 36],
    "Performance": [65, 70, 72, 78, 82, 87, 90, 94],
    "Salary": [
        45000,
        50000,
        55000,
        62000,
        70000,
        78000,
        85000,
        95000
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
x = df[[
    "Experience",
    "Age",
    "Performance"
]]

y = df[
    "Salary"
]
```

```python
print(x.shape)
print(y.shape)
```

```python
print(x.head())
print(y.head())
```

```python
print(type(x))
print(type(y))
```

```python
x_single = x[["Experience"]]
y_single = y
```

```python
x_two = x[["Age", "Performance"]]
y_two = y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
predictions = model.predict(x_test)
```

```python
print("Actual:", y_test.values)
print("Predicted:", predictions)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, predictions)

print("MAE: ", mae)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, predictions)

print("R2:", r2)
```
