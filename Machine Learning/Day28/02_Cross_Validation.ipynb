# 🤖 Cross-Validation

```python
import pandas as pd

employees = {
    "Age": [22, 24, 27, 29, 31, 26, 35, 30, 28, 33, 25, 32, 29, 36, 27, 34],
    "Experience": [1, 2, 3, 5, 7, 4, 10, 6, 3, 8, 2, 9, 5, 11, 4, 7],
    "Performance": [
        68, 72, 78, 85,
        91, 80, 97, 88,
        75, 94, 73, 92,
        86, 98, 81, 90
    ],
    "Projects": [1, 2, 3, 4, 6, 3, 9, 5, 2, 7, 2, 8, 5, 10, 4, 7],
    "Salary": [
        45000, 49000, 56000, 67000,
        82000, 61000, 120000, 76000,
        54000, 100000, 50000, 110000,
        73000, 135000, 63000, 92000
    ]
}

df = pd.DataFrame(employees)

X = df[["Age", "Experience", "Performance", "Projects"]]
y = df["Salary"]
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="r2"
)

print(scores)
```

```python
mean_r2 = scores.mean()
print(mean_r2)
std_r2 = scores.std()
print(std_r2)
```

```python
second_scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="neg_mean_absolute_error"
)

print(second_scores)

mean_mae = -second_scores.mean()
print(mean_mae)
```

```python
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy='mean')
baseline_score = cross_val_score(
    baseline,
    X,
    y,
    cv=5,
    scoring="r2"
)

baseline_mean_r2 = baseline_score.mean()

second_baseline_scores = cross_val_score(
    baseline,
    X,
    y,
    cv=5,
    scoring="neg_mean_absolute_error"
)

baseline_mean_mae = -second_baseline_scores.mean()
```

```python
compare = pd.DataFrame([
    {
        "Model": "LinearRegression",
        "Mean R2": mean_r2,
        "Mean MAE": mean_mae
    },
    {
        "Model": "Baseline",
        "Mean R2": baseline_mean_r2,
        "Mean MAE": baseline_mean_mae
    }
])

print(compare)
```
