# 🤖 Feature Engineering

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Experience": [0, 2, 5, 8, 10],
    "Projects": [1, 3, 8, 12, 15],
    "Performance": [65, 72, 84, 91, 95],
    "Salary": [45000, 58000, 85000, 120000, 140000]
})

print(employees)
```

### 🎯 Task 1 — Projects Per Year

```python
employees["Projects_Per_Year"] = np.where(
    employees["Experience"] > 0,
    employees["Projects"] / employees["Experience"],
    0
)
```

### 🎯 Task 2 — Experience Level

```python
employees["Experience_Level"] = np.where(
    employees["Experience"] >= 7,
    "Senior",
    np.where(
        employees["Experience"] >= 3,
        "Mid",
        "Junior"
    )
)
```

### 🎯 Task 3 — Performance Score

```python
employees["Performance_Score"] = employees["Performance"] / 100
```

### 🎯 Task 4 — Workload Score

```python
employees["Workload_Score"] = employees["Projects"] + employees["Experience"]
```

### 🎯 Task 5 — Inspect
