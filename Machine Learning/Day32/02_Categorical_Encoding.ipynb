# 🤖 Categorical Encoding

```python
import pandas as pd

employees = pd.DataFrame({
    "Department": [
        "AI", "Data", "Web", "AI", "Data"
    ],
    "Education": [
        "Bachelor", "Master", "Bachelor",
        "PhD", "Master"
    ],
    "Age": [24, 29, 31, 27, 35]
})

print(employees)
```

### 🎯 Task 1 — One-Hot Encoding with Pandas

```python
department_encoded = pd.get_dummies(
    employees,
    columns=["Department"]
)

print(department_encoded)
```

### 🎯 Task 2 — Encode Multiple Columns

```python
multiple_encoded = pd.get_dummies(
    employees,
    columns=["Department", "Education"]
)

print(multiple_encoded)
```

### 🎯 Task 3 — OneHotEncoder

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore"
)

encoded_data = encoder.fit_transform(employees[["Department", "Education"]])
```

### 🎯 Task 4 — Understand the Output

```python
encoded_array = encoded_data.toarray()
print(encoded_array)

print(encoder.get_feature_names_out())
```

### 🎯 Task 5 — Convert Back to DataFrame

```python
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out()
)

print(encoded_df)
```
