# 🤖 SimpleImputer — Missing Values Inside ML Pipelines

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [22, 25, np.nan, 31, 35, 29, np.nan, 40],
    "Experience": [1, 3, 5, np.nan, 8, 4, 10, 12],
    "Department": [
        "AI", "Data", "AI", "Web",
        np.nan, "Data", "AI", "Web"
    ],
    "Education": [
        "Bachelor", "Master", "Bachelor", np.nan,
        "Master", "PhD", "Master", "Bachelor"
    ],
    "Salary": [
        45000, 60000, 72000, 85000,
        100000, 78000, 120000, 135000
    ]
})

print(df)
```

```python
print(df.isnull().sum())
```

```python
from sklearn.impute import SimpleImputer

numeric_imputer = SimpleImputer(
    strategy="median"
)

categorical_imputer = SimpleImputer(
    strategy="most_frequent"
)
```

```python
numeric_feature = [
    "Age",
    "Experience"
]

X_numeric = df[numeric_feature]

X_numeric_imputed = numeric_imputer.fit_transform(X_numeric)

print(pd.DataFrame(
    X_numeric_imputed,
    columns=numeric_feature
))

categorical_feature = [
    "Department",
    "Education"
]

X_categorical = df[categorical_feature]

X_categorical_imputed = categorical_imputer.fit_transform(X_categorical)

print(pd.DataFrame(
    X_categorical_imputed,
    columns=categorical_feature
))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numeric_feature),
        ("cat", categorical_pipeline, categorical_feature)
    ]
)
```

```python
X_transformed = preprocessor.fit_transform(df)

print(X_transformed)
```
