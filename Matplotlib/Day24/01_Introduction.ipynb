# 📊 Matplotlib

```python
import matplotlib.pyplot as plt
```

### Line Plot — plt.plot()

```python
days = [1, 2, 3, 4, 5]
sales = [100, 120, 150, 140, 180]

plt.plot(days, sales)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Daily Sales")

plt.show()
```

### Scatter Plot — plt.scatter()

```python
experience = [1, 2, 3, 4, 5]
salary = [50000, 55000, 65000, 75000, 90000]

plt.scatter(experience, salary)

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")

plt.show()
```

### Bar Chart — plt.bar()

```python
departments = ["AI", "Data", "Web"]
employees = [15, 10, 8]

plt.bar(departments, employees)

plt.xlabel("Department")
plt.ylabel("Employees")
plt.title("Employees by Department")

plt.show()
```

### Histogram — plt.hist()

```python
ages = [21, 22, 24, 25, 25, 26, 27, 28, 30, 31]

plt.hist(ages)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

### Multiple Lines

```python
months = [1, 2, 3, 4]
python = [70, 75, 82, 88]
ml = [65, 72, 80, 90]

plt.plot(months, python, label="Python")
plt.plot(months, ml, label="ML")

plt.xlabel("Months")
plt.ylabel("Scores")
plt.title("Students Progress")

plt.legend()

plt.show()
```

### Practice

```python
import pandas as pd

employees = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["AI", "Data", "AI", "Web", "Data", "AI", "Web", "AI"],
    "Age": [24, 28, 26, 30, 25, 29, 27, 31],
    "Salary": [90000, 70000, 85000, 95000, 65000, 92000, 72000, 98000],
    "Experience": [2, 4, 3, 6, 1, 5, 4, 7],
    "Performance": [92, 81, 88, 95, 79, 90, 84, 96]
}

df = pd.DataFrame(employees)
```

```python
plt.plot(df.index, df["Salary"])

plt.xlabel("Employees Index")
plt.ylabel("Salary")
plt.title("Employee index vs Salary")

plt.show()
```

```python
plt.scatter(df["Experience"], df["Salary"])

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")

plt.show()
```

```python
counts = df["Department"].value_counts()
counts.plot(kind="bar")

plt.xlabel("Departments")
plt.ylabel("Number of Employees")
plt.title("Department vs Number of Employees")

plt.show()
```

```python
plt.hist(df["Salary"])

plt.xlabel("Salary")
plt.ylabel("Frequency of Salary")
plt.title("Salary Distribution")

plt.show()
```

```python
plt.scatter(df["Age"], df["Performance"])

plt.xlabel("Age")
plt.ylabel("Perofrmance")
plt.title("Performance vs Age")

plt.show()
```

```python
average_salary = (
    df.groupby("Department", as_index=False)["Salary"]
      .mean()
)

average_salary.plot(
    x="Department",
    y="Salary",
    kind="bar"
)
```

```python
plt.scatter(df["Experience"], df["Performance"])

plt.xlabel("Experience")
plt.ylabel("Performance")
plt.title("Experience vs Performance")

plt.show()
```
