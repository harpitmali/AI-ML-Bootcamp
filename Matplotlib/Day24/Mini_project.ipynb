```python
import pandas as pd
import matplotlib.pyplot as plt

employees = {
    "Name": [
        "Alice", "Bob", "Charlie", "David",
        "Eva", "Frank", "Grace", "Henry",
        "Ivy", "Jack", "Kate", "Leo"
    ],
    "Department": [
        "AI", "Data", "AI", "Web",
        "Data", "AI", "Web", "AI",
        "Data", "Web", "AI", "Data"
    ],
    "Age": [23, 25, 29, 31, 27, 24, 35, 30, 26, 33, 28, 32],
    "Experience": [1, 2, 5, 7, 4, 2, 9, 6, 3, 8, 5, 7],
    "Salary": [
        55000, 60000, 85000, 100000,
        75000, 58000, 130000, 95000,
        70000, 110000, 90000, 98000
    ],
    "Performance": [
        72, 76, 88, 94,
        82, 74, 97, 91,
        80, 96, 89, 93
    ]
}

df = pd.DataFrame(employees)
```

```python
print(df["Salary"].mean())
print(df["Salary"].median())
print(df["Salary"].max())
print(df["Salary"].min())
```

```python
print(df["Experience"].corr(df["Salary"]))
```

```python
print(df["Experience"].corr(df["Performance"]))
```

```python
plt.hist(df["Salary"])

plt.xlabel("Salary")
plt.ylabel("Frequency of Salary")
plt.title("Salary Distribution")

plt.show()
```

```python
plt.scatter(df["Experience"], df["Salary"])

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")

plt.show()
```

```python
plt.scatter(df["Experience"], df["Performance"])

plt.xlabel("Experience")
plt.ylabel("Performance")
plt.title("Experience vs Performance")

plt.show()
```

```python
average_salary = df.groupby("Department")["Salary"].mean()

average_salary.plot(kind="bar")

plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.title("Average Salary by Department")

plt.show()
```

```python
average_performance = df.groupby("Department")["Performance"].mean()

average_performance.plot(kind="bar")

plt.xlabel("Department")
plt.ylabel("Average Performance")
plt.title("Average Performace by Department")

plt.show()
```
