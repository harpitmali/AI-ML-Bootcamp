# Advanced EDA: Box Plots, Outliers & Correlation Heatmaps

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

employees = {
    "Name": [
        "Alice", "Bob", "Charlie", "David",
        "Eva", "Frank", "Grace", "Henry",
        "Ivy", "Jack", "Kate", "Leo"
    ],
    "Department": [
        "AI", "Data", "AI", "Web",
        "Data", "AI", "Web", "AI",
        "Data", "Web", "AI", "Data"
    ],
    "Age": [23, 25, 29, 31, 27, 24, 35, 30, 26, 33, 28, 32],
    "Experience": [1, 2, 5, 7, 4, 2, 9, 6, 3, 8, 5, 7],
    "Salary": [
        55000, 60000, 85000, 100000,
        75000, 58000, 130000, 95000,
        70000, 110000, 90000, 98000
    ],
    "Performance": [
        72, 76, 88, 94,
        82, 74, 97, 91,
        80, 96, 89, 93
    ]
}

df = pd.DataFrame(employees)
```

```python
plt.boxplot(df["Salary"])

plt.ylabel("Salary")
plt.title("Salary Distribution")

plt.show()
```

```python
plt.boxplot(df["Experience"])

plt.ylabel("Experience")
plt.title("Experience Distribution")

plt.show()
```

```python
plt.boxplot([
    df["Age"],
    df["Experience"],
    df["Performance"]
])

plt.xticks(
    [1,2,3],
    ["Age","Experience","Performance"]
)
plt.title("Feature Distribution")

plt.show()
```

```python
correlation = df.corr(numeric_only=True)
```

```python
correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.boxplot(df["Salary"])

plt.ylabel("Salary")
plt.title("Salary Distribution")

plt.show()
```

```python
plt.hist(df["Salary"])

plt.xlabel("Salary")
plt.ylabel("Frequency of Salary")
plt.title("Salary Distribution")

plt.show()
```
