# EDA Mini-Project — Employee Performance Analysis

### Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

employees = {
    "Name": [
        "Aarav", "Meera", "Rohan", "Priya", "Arjun",
        "Ananya", "Kabir", "Ishita", "Rahul", "Sneha",
        "Vikram", "Diya", "Karan", "Pooja", "Nikhil"
    ],

    "Department": [
        "AI", "Data", "AI", "Web", "Data",
        "AI", "Web", "Data", "AI", "Web",
        "AI", "Data", "Web", "AI", "Data"
    ],

    "Age": [
        23, 27, 31, 25, 29,
        24, 28, 26, 30, 22,
        33, 27, 32, 26, 29
    ],

    "Experience": [
        1, 3, 7, 2, 5,
        2, 4, 6, 8, 1,
        10, 3, 7, 4, 5
    ],

    "Salary": [
        55000, 68000, 92000, 62000, 85000,
        58000, 72000, 95000, 110000, 52000,
        145000, 78000, 105000, 88000, 90000
    ],

    "Performance": [
        72, 76, 91, 84, 88,
        74, 79, 93, 97, 70,
        98, 82, 95, 89, 90
    ],

    "Projects": [
        1, 2, 5, 2, 4,
        2, 3, 6, 8, 1,
        10, 3, 7, 4, 5
    ]
}

df = pd.DataFrame(employees)

print(df)
```

### Task 6 — Department Analysis

```python
department_analysis = df.groupby("Department").agg(
    Employee_Count=("Name","count"),
    Average_Salary=("Salary","mean"),
    Average_Performance=("Performance","mean"),
    Average_Experience=("Experience","mean")
)

department_analysis = department_analysis.sort_values(
    "Average_Salary",
    ascending=False
)

department_analysis
```

### Task 7 — Department Visualizations

```python
plt.bar(department_analysis.index, department_analysis["Average_Salary"])

plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Average Salary by Department")

plt.show()
```

```python
plt.bar(department_analysis.index, department_analysis["Average_Performance"])

plt.xlabel("Department")
plt.ylabel("Performance")
plt.title("Average Performance by Department")

plt.show()
```

```python
plt.bar(department_analysis.index, department_analysis["Employee_Count"])

plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.title("Employee Count by Department")

plt.show()
```

### Task 10 — High Performers

```python
high_performers = df[(df["Performance"] >= 90) & (df["Experience"] >= 4)]

high_performers = high_performers.sort_values(
    "Performance",
    ascending=False
)

high_performers
```

### Task 12 — 🔥 Final EDA Summary

```python
summary = df.groupby("Department").agg(
    Employee_Count=("Name","count"),
    Average_Salary=("Salary","mean"),
    Median_Salary=("Salary","median"),
    Average_Performance=("Performance","mean"),
    Average_Experience=("Experience","mean"),
    Maximum_Salary=("Salary","max")
)

summary = summary.sort_values(
    "Average_Salary",
    ascending=False
)

summary
```
