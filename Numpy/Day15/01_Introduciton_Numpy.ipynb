##### Experiment - 1 - Creating Your First Array

```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

##### Experiment 2 — Check the Type

```python
print(type(arr))
```

##### Experiment 3 — Creating Different Arrays

```python
arr1 = np.array((1, 2, 3, 4))

print(arr1)
```

```python
arr2 = np.array([1.2, 3.4, 5.6])

print(arr2)
```

```python
arr3 = np.array(["AI", "ML", "Python"])

print(arr3)
```

##### Experiment 4 — Array Properties

```python
arr = np.array([10, 20, 30, 40, 50])

print(f"Shape: {arr.shape}")

print(f"N-Dimension: {arr.ndim}")

print(f"Size: {arr.size}")

print(f"Data type: {arr.dtype}")
```

##### Experiment 5 — Two-Dimensional Array

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
```

```python
print(f"Shape: {matrix.shape}")

print(f"N-Dimension: {matrix.ndim}")

print(f"Size: {matrix.size}")

print(f"Data type: {matrix.dtype}")
```

##### Experiment 6 — Mixed Data Types

```python
arr = np.array([1, 2.3, 4])

print(arr)

print(arr.dtype)
```
