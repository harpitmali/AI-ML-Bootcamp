# Mini Project - Student Performance Dataset

```python
import numpy as np

students = np.array([
    [101, 85, 90, 78],
    [102, 92, 88, 95],
    [103, 70, 75, 80],
    [104, 88, 82, 91],
    [105, 76, 85, 79]
])
```

### Part 1 — Dataset Information ⭐

```python
print(students.shape)
print(students.ndim)
print(students.size)
print(students.dtype)
```

### Part 2 — Extract Individual Columns ⭐⭐

```python
student_ids = np.array([students[:,0]])
math_marks = np.array([students[:,1]])
python_marks = np.array([students[:,2]])
ml_marks = np.array([students[:,3]])
```

```python
print(f"Student IDs:\n{student_ids}\n")
print(f"Math:\n{math_marks}\n")
print(f"Python:\n{python_marks}\n")
print(f"ML:\n{ml_marks}\n")
```

### Part 3 — Individual Student Reports ⭐⭐

```python
print(f"Student 102:\n{students[1, :]}\n")
print(f"Student 105:\n{students[-1, :]}\n")
```

### Part 4 — Subject Selection ⭐⭐⭐

```python
only_marks = students[:,1:]

print(only_marks)
```

### Part 5 — Dataset Slicing ⭐⭐⭐

```python
print("First three students")
print(students[:3, :])

print("Last two students")
print(students[-2:, :])

print("Every alternate student")
print(students[::2])
```

### Part 6 — Advanced Slicing ⭐⭐⭐⭐

```python
print("Math and Python marks only for the first three students")
print(students[:3, 1:3])

print("Python and ML marks for the last two students")
print(students[-2:, 2:])
```

### Part 7 — Find Specific Values ⭐⭐⭐⭐

```python
print("Math marks of Student 104")
print(students[-2, 1])

print("Python marks of Student 103")
print(students[2, 2])

print("ML marks of Student 102")
print(students[1, 3])

print("ID of the final student")
print(students[-1, 0])
```

### 🌟 Bonus — Build a Smaller Dataset

```python
bonus = students[1:-1, [0,2,3]]

print(bonus)
```
