### Q1

A NumPy array of your favorite 5 numbers.

Print:

array
shape
ndim
size
dtype

```python
import numpy as np

arr = np.array([7, 9, 17, 21, 30])

print(arr)
```

```python
print(arr.shape)

print(arr.ndim)

print(arr.size)

print(arr.dtype)
```

### Q2

A 2×3 matrix of student marks.

Example:

90 85 88
75 80 92

Print:

matrix
shape

```python
matrix = np.array([
    [90, 85, 88],
    [75, 80, 92]
])

print(matrix)
```

```python
print(matrix.shape)

print(matrix.ndim)

print(matrix.size)

print(matrix.dtype)
```
