# Numpy Array Operations

##### Experiment 1 — Array + Scalar

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr + 5)

print(arr)
```

##### Experiment 2 — Subtraction

```python
print(arr - 5)
```

##### Experiment 3 — Multiplication

```python
arr = np.array([2, 4, 6, 8])

print(arr * 2)
```

##### Experiment 4 — Division

```python
arr = np.array([10, 20, 30, 40])

print(arr / 10)
```

##### Experiment 5 — Array + Array

```python
arr1 = np.array([10,20,30])

arr2 = np.array([1,2,3])

print(arr1 + arr2)
```

##### Experiment 6 — Array - Array

```python
arr1 = np.array([10,20,30])

arr2 = np.array([1,2,3])

print(arr1 - arr2)
```

##### Experiment 7 — Array × Array

```python
arr1 = np.array([2,4,6])

arr2 = np.array([3,5,7])

print(arr1 * arr2)
```

##### Experiment 8 — Array ÷ Array

```python
arr1 = np.array([20,30,40])

arr2 = np.array([2,3,4])

print(arr1 / arr2)
```

##### Experiment 9 — Comparison Operations

```python
arr = np.array([10,20,30,40])

print(arr > 20)

print(arr == 20)

print(arr <= 30)
```

##### Experiment 10 — Practical Example

```python
marks = np.array([45, 78, 90, 33, 65])

print(marks >= 40)

passed_students = marks[marks >= 40]

print(passed_students)
```
