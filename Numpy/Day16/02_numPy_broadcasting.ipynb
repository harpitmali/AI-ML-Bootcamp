# NumPy Broadcasting

##### Experiment 1 — Scalar Broadcasting

```python
import numpy as np

arr = np.array([10,20,30,40])

print(arr + 100)
```

##### Experiment 2 — Multiplication

```python
arr = np.array([2,4,6,8])

print(arr * 10)
```

##### Experiment 3 — 2D Broadcasting

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

print(matrix + 10)
```

##### Experiment 4 — Row Broadcasting

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

row = np.array([10,20,30])

print(matrix + row)
```

##### Experiment 5 — Column Broadcasting

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

column = np.array([
    [100],
    [200]
])

print(matrix + column)
```

##### Experiment 6 — When Broadcasting Fails

```python

arr1 = np.array([1,2,3])

arr2 = np.array([10,20])
try:
    print(arr1 + arr2)
except ValueError:
    print("Different incompatible shapes")
```
