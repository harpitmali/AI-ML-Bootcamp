# Universal Functions

##### Experiment 1 — Square Root

```python
import numpy as np

arr = np.array([1, 4, 9, 16, 25])

print(np.sqrt(arr))
```

##### Experiment 2 — Power

```python
arr = np.array([1,2,3,4])

print(np.power(arr, 2))

print(np.power(arr, 3))
```

##### Experiment 3 — Absolute Value

```python
arr = np.array([-10, 20, -30, 40])

print(np.abs(arr))
```

##### Experiment 4 — Round

```python
arr = np.array([3.14159, 2.71828, 1.23456])

print(np.round(arr, 2))
```

##### Experiment 5 — Exponential

```python
arr = np.array([1,2,3])

print(np.exp(arr))
```

##### Experiment 6 — Natural Log

```python
arr = np.array([1,2,4,8])

print(np.log(arr))
```

##### Experiment 7 — Maximum & Minimum

```python
arr = np.array([15, 80, 42, 91, 67])

print(np.max(arr))
print(np.min(arr))
```

##### Experiment 8 — Sum

```python
arr = np.array([10,20,30,40])

print(np.sum(arr))
```

##### Experiment 9 — Mean (Average)

```python
arr = np.array([80,90,100])

print(np.mean(arr))
```

##### Experiment 10 — Combined Example

```python
marks = np.array([78, 85, 92, 66, 88])

print("Highest:", np.max(marks))
print("Lowest :", np.min(marks))
print("Average:", np.mean(marks))
print("Total  :", np.sum(marks))
```
