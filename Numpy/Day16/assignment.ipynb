##### ⭐ Assignment 1 — Employee Salary Analysis

```python
import numpy as np

salary = np.array([35000, 42000, 50000, 38000, 45000])

print(salary)
print(salary + 2000)
print(np.max(salary))
print(np.min(salary))
print(np.mean(salary))
print(np.sum(salary))
```

##### ⭐⭐ Assignment 2 — Product Price Update

```python
prices = np.array([
    120,
    250,
    180,
    300,
    150
])

prices = prices * 1.10

print(prices)
print(np.round(prices,2))
print(np.max(prices))
print(np.min(prices))
```

##### ⭐⭐⭐ Assignment 3 — Student Result Analysis

```python
marks = np.array([
    35,
    82,
    67,
    91,
    48,
    76,
    29
])

print(marks >= 40)
print(marks > 80)
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
```

##### ⭐⭐⭐⭐ Assignment 4 — Sales Dataset

```python
sales = np.array([
    [120,150,180],
    [200,220,250],
    [300,330,360]
])

sales += 20
sales = sales * 1.05
print(np.max(sales))
print(np.min(sales))
print(np.mean(sales))
print(np.sum(sales))
```

##### ⭐⭐⭐⭐⭐ Assignment 5 — AI Dataset Preprocessing

```python
dataset = np.array([
    [25,50000],
    [30,70000],
    [22,45000],
    [35,80000],
    [28,65000]
])

age = dataset[:,0]
salary = dataset[:,1]

print(np.mean(age))
print(np.max(salary))
print(np.min(salary))
print(np.sum(salary))
print(age > 25)
```

##### ⭐⭐⭐⭐⭐⭐ Assignment 6 — Scientific Computation

```python
numbers = np.array([
    4,
    9,
    16,
    25,
    36
])

print(np.sqrt(numbers))
print(np.power(numbers, 3))
print(np.sum(numbers))
print(np.mean(numbers))
print(np.max(numbers))
print(np.min(numbers))
```
