# Experiments

##### ⭐ Experiment 1 — Student Marks Bonus

```python
import numpy as np

marks = np.array([78, 85, 92, 66, 88])

print(f"Original Marks:\n{marks}")
print(f"Updated Marks:\n{marks + 5}")
```

##### ⭐⭐ Experiment 2 — Monthly Expenses

```python
expenses = np.array([
    1200,
    1500,
    1800,
    1000,
    2200
])

print(expenses + 200)
print(expenses - 100)
print(expenses * 2)
print(expenses / 2)
```

##### ⭐⭐⭐ Experiment 3 — Company Sales

```python
sales = np.array([
    [120,150,180],
    [200,220,250],
    [300,330,360]
])

print(sales + 50)

bonus = np.array([10,20,30])

print(sales + bonus)
```

##### ⭐⭐⭐⭐ Experiment 4 — Student Grades

```python
marks = np.array([35,42,67,89,28,95])

print(marks >= 40)
print(marks >= 80)
print(marks < 50)
```

##### ⭐⭐⭐⭐⭐ Experiment 5 — Scientific Calculator

```python
numbers = np.array([1,4,9,16,25])

print(np.sqrt(numbers))
print(np.power(numbers, 2))
print(np.sum(numbers))
print(np.mean(numbers))
print(np.min(numbers))
print(np.max(numbers))
```

##### ⭐⭐⭐⭐⭐⭐ Experiment 6 — Temperature Analysis

```python
temperatures = np.array([
    31.6,
    29.8,
    33.2,
    30.4,
    28.9
])

print(np.round(temperatures, 1))
print(np.min(temperatures))
print(np.max(temperatures))
print(np.mean(temperatures))
```

##### ⭐⭐⭐⭐⭐⭐⭐ Experiment 7 — Bank Balance

```python
balance = np.array([
    -500,
    1200,
    -300,
    700,
    -100
])

print(np.abs(balance))
```

#### 🌟 Bonus Experiment — Mini Dataset Analysis

```python
students = np.array([
    [85,90,78],
    [92,88,95],
    [70,75,80],
    [88,82,91]
])

print(students + 5)

print(np.max(students))
print(np.min(students))
print(np.mean(students))
print(np.sum(students))
```
