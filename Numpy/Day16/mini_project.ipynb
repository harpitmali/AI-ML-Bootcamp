# Mini Project — AI Student Dataset Analyzer

##### Dataset

```python
import numpy as np

students = np.array([
    [101, 85, 90, 78],
    [102, 92, 88, 95],
    [103, 70, 75, 80],
    [104, 88, 82, 91],
    [105, 76, 85, 79]
])
```

##### ⭐ Part 1 — Dataset Information

```python
print(students.shape)
print(students.ndim)
print(students.size)
print(students.dtype)
```

##### ⭐⭐ Part 2 — Extract Columns

```python
student_ids = students[:, 0]
math_marks = students[:, 1]
python_marks = students[:, 2]
ml_marks = students[:, 3]

print(student_ids)
print(math_marks)
print(python_marks)
print(ml_marks)
```

##### ⭐⭐⭐ Part 3 — Add Bonus Marks

```python
updated_students = students.copy()

updated_students[:,1:] += 5

print(updated_students)
```

##### ⭐⭐⭐⭐ Part 4 — Dataset Statistics

```python
print("Math")
print(f"Highest: {np.max(updated_students[:, 1])}")
print(f"Lowest: {np.min(updated_students[:, 1])}")
print(f"Average: {np.mean(updated_students[:, 1])}")
print()
print("Python")
print(f"Highest: {np.max(updated_students[:, 2])}")
print(f"Lowest: {np.min(updated_students[:, 2])}")
print(f"Average: {np.mean(updated_students[:, 2])}")
print()
print("ML")
print(f"Highest: {np.max(updated_students[:, 3])}")
print(f"Lowest: {np.min(updated_students[:, 3])}")
print(f"Average: {np.mean(updated_students[:, 3])}")
```

##### ⭐⭐⭐⭐⭐ Part 5 — Student Reports

```python
print(f"Student ID :{updated_students[1,0]}")
print(f"Math: {updated_students[1, 1]}")
print(f"Python: {updated_students[1, 2]}")
print(f"ML: {updated_students[1, 3]}")
print()
print(f"Student ID :{updated_students[-1,0]}")
print(f"Math: {updated_students[-1, 1]}")
print(f"Python: {updated_students[-1, 2]}")
print(f"ML: {updated_students[-1, 3]}")
```

##### ⭐⭐⭐⭐⭐⭐ Part 6 — Performance Check

```python
print("Students scoring 90 or more in:")
print("Math")
print(updated_students[:,1] >= 90)
print()
print("Python")
print(updated_students[:,2] >= 90)
print()
print("ML")
print(updated_students[:,3] >= 90)
```

##### ⭐⭐⭐⭐⭐⭐⭐ Part 7 — Overall Dataset Summary

```python
print(f"Total Student: {updated_students.shape[0]}")
print(f"Total Subject: {updated_students.shape[1]-1}")
print(f"Highest Marks: {np.max(updated_students[:, 1:])}")
print(f"Lowest Marks: {np.min(updated_students[:, 1:])}")
print(f"Average Marks: {np.mean(updated_students[:, 1:])}")
```
