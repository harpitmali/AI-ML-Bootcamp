# Introduction to Pandas

##### Creating Series

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40])

print(s)
```

```python
marks = pd.Series(
    [97, 80, 95],
    index = ["Harpit", "Luffy", "Zoro"]
)

print(marks)
```

##### Creating DataFrame

```python
student = {
    "Name": ["Luffy", "Zoro", "Sanji"],
    "Age": [20, 24, 22],
    "Marks": [80, 85, 90]
}

df = pd.DataFrame(student)

print(df)
```

##### Creating from Numpy Array

```python
import numpy as np

arr = np.array([
    [1,2],
    [3,4],
    [5,6]
])

df = pd.DataFrame(
    arr,
    columns = ["A", "B"]
)

print(df)
```

##### Real-Life Example

```python
employees = {
    "ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"],
    "Salary": [50000, 70000, 65000]
}

df = pd.DataFrame(employees)

print(df)
```
