# 🐼 Inspecting a DataFrame

```python
import pandas as pd

students = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Harpit", "Luffy", "Zoro", "Ussop", "Sanji"],
    "Age": [21, 19, 21, 20, 22],
    "Python": [92, 85, 78, 95, 88],
    "ML": [90, 80, 82, 96, 91]
}

df = pd.DataFrame(students)
```

#### 1. head() -> Shows the first 5 rows.

```python
print(df.head())

print(df.head(3))
```

#### 2️. tail() -> Shows the last 5 rows.

```python
print(df.tail())
print()
print(df.tail(2))
```

#### 3. shape -> Tells you the dimensions of the DataFrame.

```python
print(df.shape)
```

#### 4. columns -> Displays all column names.

```python
print(df.columns)
```

#### 5. index -> Displays row labels.

```python
print(df.index)
```

#### 6. dtypes -> Shows the data type of every column.

```python
print(df.dtypes)
```

#### 7. info() -> It tells you:

Number of rows,
Number of columns,
Data types,
Missing values,
Memory usage,

```python
print(df.info())
```

#### 8. describe() -> Shows statistics for numeric columns.

```python
print(df.describe())
```
