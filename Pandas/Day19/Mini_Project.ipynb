# 🐼 Mini Project — Student Performance Analysis

```python
import pandas as pd

students = {
    "ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Harpit", "Luffy", "Zoro", "Naruto", "Sanji", "Nami"],
    "Age": [21, 19, 21, 20, 22, 20],
    "Python": [92, 85, 78, 95, 88, 91],
    "ML": [90, 80, 82, 96, 91, 89],
    "DSA": [94, 79, 85, 97, 90, 88]
}

df = pd.DataFrame(students)
```

### 🟢 Task 1 — Inspect the Dataset

```python
print(df.head())
print()

print(df.tail())
print()

print(df.shape)
print()

print(df.columns)
print()

print(df.index)
print()

df.info()
print()

print(df.describe())
print()
```

### 🟢 Task 3 — Display Specific Information

```python
print(f"Students Name\n{df['Name']}\n")

print(f"Python Marks\n{df['Python']}\n")

print(f"ML Name\n{df['ML']}\n")

print(f"DSA Name\n{df['DSA']}\n")
```

### 🟢 Task 4 — Compare Subjects

```python
print("Highest Python:", df["Python"].max())

print("Highest ML:", df["ML"].max())

print("Highest DSA:", df["DSA"].max())
```

```python
print("Average Python:", df["Python"].mean())

print("Average ML:", df["ML"].mean())

print("Average DSA:", df["DSA"].mean())
```

### 🟢 Task 5 — Overall Dataset Summary

```python
print("Total Students:", df.shape[0])

print("Total Features:", df.shape[1])

print("Highest Score:", df[["Python", "ML", "DSA"]].max().max())

print("Lowest Score:", df[["Python", "ML", "DSA"]].min().min())

print("Overall Average:", df[["Python", "ML", "DSA"]].mean().mean())
```
