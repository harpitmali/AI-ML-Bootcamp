# 🐼 Pandas Indexing & Column Selection

```python
import pandas as pd

students = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Harpit", "Luffy", "Zoro", "Naruto", "Sanji"],
    "Age": [21, 19, 21, 20, 22],
    "Python": [92, 85, 78, 95, 88],
    "ML": [90, 80, 82, 96, 91]
}

df = pd.DataFrame(students)
```

### 1️⃣ Selecting One Column

```python
df["Name"]
```

```python
df["Age"]
```

```python
df["Python"]
```

### 2️⃣ Selecting Multiple Columns

```python
df[["Name", "Python"]]
```

### 3️⃣ iloc — Integer Location

```python
print(df.iloc[0, 1])
```

```python
print(df.iloc[1])
```

```python
print(df.iloc[0:3])
```

```python
print(df.iloc[:, 1])
```

```python
print(df.iloc[:, 1:4])
```

### 4️⃣ loc — Label Location

```python
print(df.loc[1, "Name"])
```

```python
print(df.loc[:, "Python"])
```

```python
print(df.loc[:, ["Name", "ML"]])
```

```python
print(df.loc[2])
```
