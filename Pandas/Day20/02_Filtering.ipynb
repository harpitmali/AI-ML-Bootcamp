# 🐼 Filtering

```python
import pandas as pd

students = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Harpit", "Luffy", "Zoro", "Naruto", "Sanji"],
    "Age": [21, 19, 21, 20, 22],
    "Python": [92, 85, 78, 95, 88],
    "ML": [90, 80, 82, 96, 91]
}

df = pd.DataFrame(students)
```

### Boolean Mask

```python
df["Age"] > 20
```

```python
df[df["Age"] > 20]
```

```python
df[df["Age"] < 21]
```

```python
df[df["Python"] >= 90]
```

```python
df[df["ML"] <= 85]
```

```python
df[df["Age"] == 21]
```

```python
df[df["Age"] != 21]
```

### 🌟 Multiple Conditions

```python
df[(df["Age"] > 20) & (df["Python"] > 90)]
```

### OR Condition

```python
df[(df["Age"] == 19) | (df["Python"] > 90)]
```

### AND Condition

```python
df[(df["Age"] > 20) & (df["Python"] > 90)]
```
