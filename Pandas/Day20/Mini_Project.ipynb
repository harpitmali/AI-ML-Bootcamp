# 🐼 Mini Project — Employee Data Analysis

```python
import pandas as pd

employees = {
    "ID": [101,102,103,104,105,106,107],
    "Name": ["Alice","Bob","Charlie","David","Eva","Frank","Grace"],
    "Department": ["AI","Web","AI","Data","Web","AI","Data"],
    "Age": [24,28,26,30,25,29,27],
    "Salary": [90000,70000,85000,95000,65000,92000,88000],
    "Experience": [2,4,3,6,1,5,4]
}

df = pd.DataFrame(employees)
```

### 🟢 Task 1 — Inspect the Dataset

```python
print(df.head())
print()

print(df.tail())
print()

print(df.shape)
print()

print(df.columns)
print()

print(df.index)
print()

print(df.dtypes)
print()

df.info()
print()

print(df.describe())
print()
```

### 🌟 Bonus Challenge

```python
bonus = []

for salary in df["Salary"]:
    if salary >= 90000:
        bonus.append(10000)
    else:
        bonus.append(5000)

df["Bonus"] = bonus
```
