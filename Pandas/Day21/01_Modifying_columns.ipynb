# 🐼 Creating & Modifying Pandas Columns

### 1️⃣ Creating a New Column

```python
import pandas as pd
import numpy as np
 
students = {
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Harpit", "Luffy", "Zoro", "Ussop", "Sanji"],
    "Python": [92, 85, 78, 95, 88],
    "ML": [90, 80, 82, 96, 91],
    "DSA": [94, 79, 85, 97, 90]
}

df = pd.DataFrame(students)

print(df)
```

```python
df["Total"] = df["Python"] + df["ML"] + df["DSA"]

print(df)
```

### 2️⃣ Creating an Average Column

```python
df["Average"] = (
    df["Python"] +
    df["ML"] +
    df["DSA"]
) / 3

df["Average"] = df["Average"].round(2)

print(df)
```

### 3️⃣ Modifying an Existing Column

```python
df["Python"] += 5

df
```

### 4️⃣ Creating a Column Based on a Condition

```python
df["Result"] = np.where(
    df["Total"] >= 240,
    "Pass",
    "Fail"
)

print(df)
```

### 5️⃣ Multiple Conditions

```python
df["Performance"] = np.where(
    df["Total"] >= 270,
    "Excellent",
    np.where(
        df["Total"] >= 240,
        "Good",
        "Needs Improvement"
    )
)

print(df)
```

### 6️⃣ Creating a Column From Another Column

```python
df["Percentage"] = (df["Total"] / 300) * 100

df["Percentage"] = df["Percentage"].round(2)

print(df)
```

### 7️⃣ Column Order

```python
print(df.columns)
```
