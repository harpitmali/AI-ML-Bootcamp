# 🐼 Pandas Sorting Data

```python
import pandas as pd

employees = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace"],
    "Department": ["AI", "Web", "AI", "Data", "Web", "AI", "Data"],
    "Age": [24, 28, 26, 30, 25, 29, 27],
    "Salary": [90000, 70000, 85000, 95000, 65000, 92000, 88000],
    "Experience": [2, 4, 3, 6, 1, 5, 4]
}

df = pd.DataFrame(employees)

print(df)
```

### 1️⃣ Basic Sorting

```python
df.sort_values("Salary")
```

### 2️⃣ Descending Order

```python
df.sort_values("Salary", ascending=False)
```

### 3️⃣ Sorting by Another Column

```python
df.sort_values("Age")
```

```python
df.sort_values("Age", ascending=False)
```

### 4️⃣ Sorting by Multiple Columns

```python
df.sort_values(["Department", "Salary"])
```

```python
df.sort_values(["Department", "Salary"], ascending=False)
```

### 5️⃣ Different Directions for Different Columns

```python
df.sort_values(
    ["Department", "Salary"], 
    ascending=[True, False]
)
```

### 8️⃣ Sorting After Filtering

```python
df[
    (df["Department"] == "AI") & 
    (df["Salary"] > 80000)
].sort_values(
    "Salary",
    ascending=False
)
```

### 9️⃣ inplace=True

```python
df.sort_values("Salary", ascending=False, inplace=True)

df
```

```python
df.sort_values("ID", inplace=True)

df
```

### Bonus

```python
df.sort_values(
    "Experience",
    ascending=False
)[["Name", "Department", "Experience", "Salary"]].head(3)
```
