# 🐼 Pandas EDA Assignment — Employee Analysis

### 📊 Dataset

```python
import pandas as pd
import numpy as np

employees = {
    "ID": [201, 202, 203, 204, 205, 206, 207, 208],
    "Name": [
        "Aarav", "Meera", "Rohan", "Priya",
        "Arjun", "Ananya", "Kabir", "Ishita"
    ],
    "Department": [
        "AI", "Data", "Web", "AI",
        "Data", "AI", "Web", "Data"
    ],
    "Age": [23, 27, 31, 25, 29, 24, 28, 26],
    "Salary": [75000, 85000, 95000, 88000, 92000, 78000, 72000, 90000],
    "Experience": [1, 3, 7, 2, 5, 2, 4, 3],
    "Performance": [82, 91, 88, 95, 89, 84, 79, 93]
}

df = pd.DataFrame(employees)

print(df)
```

### 🟢 Part 2 — Basic Analysis

```python
average_salary = np.mean(df["Salary"])

print(average_salary)
```

```python
highest_salary = np.max(df["Salary"])

print(highest_salary)
```

```python
lowest_salary = np.min(df["Salary"])

print(lowest_salary)
```

```python
average_experience = np.mean(df["Experience"])

print(average_experience)
```

```python
average_performance = np.mean(df["Performance"])

print(average_performance)
```

### 🟢 Part 3 — Filtering

```python
df[df["Salary"] > 85000]
```

```python
df[df["Department"] == "AI"]
```

```python
df[df["Experience"] >= 4]
```

```python
df[(df["Department"] == "AI") & (df["Salary"] > 80000)]
```

```python
df[(df["Age"] < 27) & (df["Performance"] >= 90)]
```

### 🟢 Part 4 — Create New Columns

```python
df["Annual_Bonus"] = np.where(
    df["Salary"] >= 90000,
    10000,
    5000
)

df
```

```python
df["Total_Compensation"] = df["Salary"] + df["Annual_Bonus"]

df
```

```python
df["Performance_Level"] = np.where(
    df["Performance"] >= 90,
    "Excellent",
    np.where(
        df["Performance"] >= 80,
        "Good",
        "Needs Improvement"
    )
)

df
```

### 🟢 Part 5 — Sorting

```python
df.sort_values(
    "Salary",
    ascending=False
).head(3)
```

```python
df.sort_values(
    "Performance",
    ascending=False
).head(3)
```

```python
df.sort_values(
    ["Department", "Performance"],
    ascending=[True, False]
)
```

### 🟢 Part 6 — Combined EDA Questions

```python
df[df["Department"] == "AI"].sort_values(
    "Salary",
    ascending=False
).head(1)
```

```python
df[(df["Experience"] >= 3) & (df["Performance"] >= 90)].sort_values(
    "Performance",
    ascending=False
)
```

```python
df.sort_values(
    "Total_Compensation",
    ascending=False
).head(1)
```

```python
print(df[df["Department"] == "AI"]["Salary"].mean())
```

```python
total_employees = df[(df["Salary"] >= 85000) & (df["Performance"] >= 85)].sort_values(
    "Performance",
    ascending=False
)[["Name", "Department", "Salary", "Performance", "Total_Compensation"]]

total_employees
```
