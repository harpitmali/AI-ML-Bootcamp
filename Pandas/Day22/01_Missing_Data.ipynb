# 🐼 Missing Data

### 1️⃣ Create a Dataset

```python
import pandas as pd
import numpy as np

employees = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [24, np.nan, 26, 30, np.nan, 29],
    "Department": ["AI", "Data", np.nan, "AI", "Web", "Data"],
    "Salary": [90000, 75000, np.nan, 95000, 65000, np.nan],
    "Experience": [2, 4, 3, np.nan, 1, 5]
}

df = pd.DataFrame(employees)

print(df) 
```

### 2️⃣ Detect Missing Values

```python
df.isnull()
```

### 3️⃣ Count Missing Values

```python
df.isnull().sum()
```

### 4️⃣ Find Rows Containing Missing Values

```python
df[df.isnull().any(axis=1)]
```

### 5️⃣ Find Rows With No Missing Values

```python
df[df.notnull().all(axis=1)]
```

### 6️⃣ Dropping Missing Values

```python
df.dropna()
```

### 7️⃣ Filling Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df
```

```python
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

df
```

```python
df.isnull().sum()
```

```python
df["Experience"] = df["Experience"].fillna(df["Experience"].mean())

df
```

```python
df.isnull().sum()
```
