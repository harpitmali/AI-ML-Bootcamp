# 🐼 Handling Missing Data in Pandas

```python
import pandas as pd
import numpy as np

employees = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [24, np.nan, 26, 30, np.nan, 29],
    "Department": ["AI", "Data", np.nan, "AI", "Web", "Data"],
    "Salary": [90000, 75000, np.nan, 95000, 65000, np.nan],
    "Experience": [2, 4, 3, np.nan, 1, 5]
}

df = pd.DataFrame(employees)

print(df)
```

### 1️⃣ dropna() — Remove Missing Rows

```python
df.dropna()
```

### 2️⃣ Drop Rows Only When a Specific Column Is Missing

```python
df.dropna(subset="Salary")
```

### 3️⃣ fillna() — Replace Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df
```

### 4️⃣ Median Instead of Mean

```python
df["Salary"].fillna(df["Salary"].median())
```

### 5️⃣ Forward Fill

```python
df["Salary"].ffill()
```

### 6️⃣ Backward Fill

```python
df["Salary"].bfill()
```

```python
df["Department"].fillna(
    df["Department"].mode()[0]
)
```
