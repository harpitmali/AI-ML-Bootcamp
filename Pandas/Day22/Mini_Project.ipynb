# 🐼 Pandas Data-Cleaning Mini Project

### Dataset

```python
import pandas as pd
import numpy as np

employees = {
    "ID": [301, 302, 303, 304, 305, 306, 307, 308, 309, 310],
    "Name": [
        "Aarav", "Meera", "Rohan", "Priya", "Arjun",
        "Ananya", "Kabir", "Ishita", "Rahul", "Sneha"
    ],
    "Department": [
        "AI", "Data", "AI", np.nan, "Web",
        "AI", "Data", "Data", np.nan, "AI"
    ],
    "Age": [
        24, 27, np.nan, 31, 26,
        np.nan, 29, 25, 28, 23
    ],
    "Salary": [
        78000, 85000, 92000, np.nan, 68000,
        88000, np.nan, 76000, 150000, 82000
    ],
    "Experience": [
        1, 3, 5, 7, np.nan,
        2, 4, np.nan, 6, 1
    ],
    "Performance": [
        82, 91, 88, 95, np.nan,
        84, 79, 93, 97, np.nan
    ]
}

df = pd.DataFrame(employees)

print(df)
```

### 🔧 Part 1 — Data Cleaning

```python
clean_df = df.copy()

clean_df
```

```python
clean_df["Department"] = clean_df["Department"].fillna(
    clean_df["Department"].mode()[0]
)
```

```python
clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].mean()
)
```

```python
clean_df["Salary"] = clean_df["Salary"].fillna(
    clean_df["Salary"].median()
)
```

```python
clean_df["Experience"] = clean_df["Experience"].fillna(
    clean_df["Experience"].mean()
)
```

```python
clean_df["Performance"] = clean_df["Performance"].fillna(
    clean_df["Performance"].mean()
)
```

```python
clean_df.isnull().sum()
```

### 🧮 Part 2 — Feature Engineering

```python
clean_df["Annual_Bonus"] = np.where(
    clean_df["Salary"] >= 90000,
    10000,
    5000
)

clean_df.head()
```

```python
clean_df["Total_Compensation"] = clean_df["Salary"] + clean_df["Annual_Bonus"]

clean_df.head()
```

```python
clean_df["Performance_Level"] = np.where(
    clean_df["Performance"] >= 90,
    "Excellent",
    np.where(
        clean_df["Performance"] >= 80,
        "Good",
        "Needs Improvement"
    )
)

clean_df.head()
```

### 🔎 Part 3 — Data Analysis

```python
clean_df.sort_values(
    "Salary",
    ascending=False
)[["Name", "Department", "Salary", "Experience"]].head(5)
```

```python
clean_df.sort_values(
    "Performance",
    ascending=False
)[["Name", "Performance", "Department"]].head(3)
```

### 🔥 Part 5 — Final Challenge

```python
top_performers = clean_df[
    (clean_df["Salary"] >= 85000) & 
    (clean_df["Performance"] >= 85) & 
    (clean_df["Experience"] >= 3)
].sort_values(
    "Performance",
    ascending=False
)[[
    "Name", 
    "Department", 
    "Salary", 
    "Experience", 
    "Performance", 
    "Performance_Level", 
    "Total_Compensation"
]]

top_performers
```
