# 🐼 Pandas: groupby() + Aggregation

### 1️⃣ What is groupby()?

```python
import pandas as pd

employees = {
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Department": ["AI", "AI", "Data", "Data", "Web", "Web"],
    "Salary": [80000, 90000, 70000, 85000, 60000, 75000],
    "Performance": [85, 95, 80, 90, 75, 88]
}

df = pd.DataFrame(employees)
```

```python
df.groupby("Department")
```

### 2️⃣ groupby() + mean()

```python
df.groupby("Department")["Salary"].mean()
```

### 3️⃣ Other aggregation functions

```python
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].min()
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")["Salary"].count()
```

### 4️⃣ Grouping without selecting a column

```python
df.groupby("Department").mean(numeric_only=True)
```

### 5️⃣ agg() — Multiple calculations at once

```python
df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min"]
)
```

### 6️⃣ Different aggregations for different columns

```python
df.groupby("Department").agg({
    "Salary": ["mean", "max"],
    "Performance": ["mean", "max", "min"]
})
```

### 7️⃣ groupby() with multiple columns

```python
df.groupby(["Department", "Performance"])
```

### 8️⃣ Getting a normal DataFrame back

```python
result = df.groupby("Department")["Salary"].mean()


result = result.reset_index()

result
```
