# 🐼 Pandas: value_counts(), unique(), nunique()

### 1. value_counts()

```python
import pandas as pd
import numpy as np

employees = {
    "ID": [401, 402, 403, 404, 405, 406, 407, 408, 409, 410, 411, 412],
    
    "Name": [
        "Aarav", "Meera", "Rohan", "Priya",
        "Arjun", "Ananya", "Kabir", "Ishita",
        "Rahul", "Sneha", "Vikram", "Diya"
    ],
    
    "Department": [
        "AI", "Data", "AI", "Web",
        "Data", "AI", "Web", "Data",
        "AI", "Web", "AI", "Data"
    ],
    
    "City": [
        "Delhi", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Pune", "Bangalore", "Mumbai"
    ],
    
    "Age": [
        24, 27, 31, 25,
        29, 23, 28, 26,
        30, 22, 25, 28
    ],
    
    "Salary": [
        78000, 85000, 92000, 68000,
        91000, 88000, 72000, 95000,
        105000, 70000, 89000, 87000
    ],
    
    "Performance": [
        82, 91, 88, 95,
        89, 84, 79, 93,
        97, 81, 90, 86
    ]
}

df = pd.DataFrame(employees)

print(df)
```

### 1. value_counts()

```python
df["Department"].value_counts()
```

### 2. unique()

```python
print(df["Department"].unique())
```

### 3. nunique()

```python
df["Department"].nunique()
```

### 4. value_counts(normalize=True)

```python
df["Department"].value_counts(normalize=True)
```

```python
df["Department"].value_counts(normalize=True) * 100
```

### 7. value_counts() on numerical columns

```python
df["Age"].value_counts()
```
