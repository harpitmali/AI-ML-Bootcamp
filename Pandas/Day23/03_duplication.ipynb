# 🐼 Pandas: Duplicates + replace() + rename()

```python
import pandas as pd
import numpy as np

employees = {
    "Employee_ID": [
        501, 502, 503, 504, 505,
        506, 507, 508, 509, 510,
        510, 503
    ],

    "Employee_Name": [
        "Aarav", "Meera", "Rohan", "Priya", "Arjun",
        "Ananya", "Kabir", "Ishita", "Rahul", "Sneha",
        "Sneha", "Rohan"
    ],

    "Department": [
        "AI", "Data", "AI", "Artificial Intelligence",
        "Data", "AI", "Web", "Web",
        "AI", "Data", "Data", "AI"
    ],

    "City": [
        "Delhi", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Pune", "Pune", "Delhi"
    ],

    "Salary": [
        78000, 85000, 92000, 68000,
        91000, 88000, 72000, 95000,
        105000, 70000, 70000, 92000
    ],

    "Performance": [
        82, 91, 88, 95,
        89, 84, 79, 93,
        97, 81, 81, 88
    ]
}

df = pd.DataFrame(employees)

print(df)
```

### 1. duplicated()

```python
df.duplicated()
```

```python
print(df.duplicated().sum())
```

### 2. drop_duplicates()

```python
df.drop_duplicates()
```

### 3. Duplicates based on a specific column

```python
df.duplicated(subset=["Employee_ID"])
```

### 4. replace()

```python
df["Department"] = df["Department"].replace(
    "Artificial Intelligence",
    "AI"
)

df
```

### 5. rename()

```python
df = df.rename(columns={
    "Employee_ID": "ID",
    "Employee_Name": "Name"
})

df
```

```python
df = df.drop_duplicates()
```
