# 🏢 Mini-Project: Employee Data Analysis

### Create the dataset

```python
import pandas as pd
import numpy as np

employees = {
    "Employee_ID": [
        601, 602, 603, 604, 605,
        606, 607, 608, 609, 610,
        610, 603
    ],

    "Name": [
        "Aarav", "Meera", "Rohan", "Priya", "Arjun",
        "Ananya", "Kabir", "Ishita", "Rahul", "Sneha",
        "Sneha", "Rohan"
    ],

    "Department": [
        "AI", "Data", "AI", "Artificial Intelligence",
        "Data", "AI", "Web", "Web",
        np.nan, "Data", "Data", "AI"
    ],

    "City": [
        "Delhi", "Mumbai", "Delhi", "Pune",
        "Mumbai", "Delhi", "Pune", "Mumbai",
        "Delhi", "Pune", "Pune", "Delhi"
    ],

    "Age": [
        24, 27, np.nan, 25, 29,
        23, 28, 26, 30, np.nan,
        np.nan, 31
    ],

    "Salary": [
        78000, 85000, 92000, np.nan,
        91000, 88000, 72000, 95000,
        105000, 70000, 70000, 92000
    ],

    "Experience": [
        1, 3, 7, 2, np.nan,
        2, 4, 6, 8, 1,
        1, 7
    ],

    "Performance": [
        82, 91, 88, 95, 89,
        84, 79, 93, 97, np.nan,
        np.nan, 88
    ]
}

df = pd.DataFrame(employees)

print(df)
```

### Part 1 — Create the cleaning pipeline

```python
clean_df = df.copy()
```

### Part 2 — Data Cleaning

```python
clean_df.isnull()
```

```python
clean_df["Department"] = clean_df["Department"].fillna(
    clean_df["Department"].mode()[0]
)
```

```python
clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].mean()
)
```

```python
clean_df["Salary"] = clean_df["Salary"].fillna(
    clean_df["Salary"].median()
)
```

```python
clean_df["Experience"] = clean_df["Experience"].fillna(
    clean_df["Experience"].mean()
)
```

```python
clean_df["Performance"] = clean_df["Performance"].fillna(
    clean_df["Performance"].mean()
)
```

```python
clean_df.isnull()
```

```python
clean_df.duplicated()
```

```python
clean_df["Department"] = clean_df["Department"].replace(
    "Artificial Intelligence",
    "AI"
)
```

```python
clean_df = clean_df.rename(columns={
    "Employee_ID": "ID"
})
```

```python
clean_df = clean_df.drop_duplicates(
    subset="ID",
    keep="first"
)
```

```python
clean_df
```

```python
clean_df["Annual_Bonus"] = np.where(
    clean_df["Salary"] >= 90000,
    10000,
    5000
)

clean_df["Total_Compensation"] = clean_df["Salary"] + clean_df["Annual_Bonus"]

clean_df
```

```python
clean_df["Performance_Level"] = np.where(
    clean_df["Performance"] >= 90,
    "Excellent",
    np.where(
        clean_df["Performance"] >= 80,
        "Good",
        "Needs Improvement"
    )
)

clean_df
```

```python
result = clean_df.groupby("Department")["Salary"].mean()
result = result.reset_index()

result
```

```python
result.sort_values(
    "Salary",
    ascending=False
).head(1)
```

```python
result = clean_df.groupby("Department")["Performance"].mean()
result = result.reset_index()

result
```

```python
result.sort_values(
    "Performance",
    ascending=False
).head(1)
```

```python
summary = clean_df.groupby("Department").agg(
    Employee_Count=("ID", "count"),
    Average_Salary=("Salary", "mean"),
    Average_Performance=("Performance", "mean"),
    Maximum_Salary=("Salary", "max")
)

summary = summary.sort_values(
    "Average_Salary",
    ascending=False
)

summary
```

```python
final_summary = clean_df.sort_values(
    "Total_Compensation",
    ascending=False
)[[
    "ID",
    "Name",
    "Department",
    "City",
    "Age",
    "Salary",
    "Experience",
    "Performance",
    "Annual_Bonus",
    "Total_Compensation",
    "Performance_Level"
]]

final_summary
```
