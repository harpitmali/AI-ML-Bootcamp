# 🐼 Final Pandas Tools: merge(), concat(), CSV

```python
import pandas as pd

employees = pd.DataFrame({
    "ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"]
})

salary = pd.DataFrame({
    "ID": [101, 102, 103],
    "Salary": [90000, 70000, 85000]
})

employees2 = pd.DataFrame({
    "ID": [104, 105],
    "Name": ["David", "Eva"]
})
```

```python
merged = employees.merge(
    salary,
    on="ID"
)
merged
```

```python
merged = pd.concat(
    [merged, employees2],
    ignore_index=True
)
merged
```

```python
merged.to_csv("employees_with_salary.csv", index=False)
```
