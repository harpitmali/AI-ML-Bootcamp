# 🎯 Mini Integration Project

### Dataset 1 — Employee Information

```python
import pandas as pd
import numpy as np

employees = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 106],
    "Name": [
        "Alice", "Bob", "Charlie",
        "David", "Eva", "Frank"
    ],
    "Department": [
        "AI", "Data", "AI",
        "Web", "Data", "AI"
    ],
    "Age": [24, 28, 26, 30, 25, 29]
})
```

### Dataset 2 — Salary Information

```python
salary = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 106],
    "Salary": [
        90000, 70000, 85000,
        95000, 65000, 92000
    ],
    "Experience": [2, 4, 3, 6, 1, 5]
})
```

### Dataset 3 — Performance Information

```python
performance = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 106],
    "Performance": [92, 81, 88, 95, 79, 90]
})
```

### 🟢 Task 1 — Merge the Data

```python
merged = employees.merge(
    salary,
    on="ID"
)
```

```python
merged = merged.merge(
    performance,
    on="ID"
)
```

### 🟢 Task 2 — Inspect

```python
merged.head()
```

```python
print(merged.shape)
```

```python
print(merged.dtypes)
```

### 🟢 Task 3 — Create Features

```python
merged["Annual_Bonus"] = np.where(
    merged["Salary"] >= 90000,
    10000,
    5000
)

merged["Total_Compensation"] = merged["Salary"] + merged["Annual_Bonus"]

merged["Performance_Level"] = np.where(
    merged["Performance"] >= 90,
    "Excellent",
    np.where(
        merged["Performance"] >= 80,
        "Good",
        "Needs Improvement"
    )
)
```

```python
merged
```

### 🟢 Task 4 — Filtering

```python
merged[(merged["Salary"] >= 85000) & (merged["Performance"] >= 85)]
```

### 🟢 Task 5 — Top Employees

```python
merged.sort_values(
    "Total_Compensation",
    ascending=False
)[[
    "Name",
    "Department",
    "Salary",
    "Total_Compensation"
]].head(3)
```

### 🟢 Task 6 — Department Analysis

```python
summary = merged.groupby("Department").agg(
    Employee_Count=("ID", "count"),
    Average_Salary=("Salary", "mean"),
    Average_Performance=("Performance", "mean"),
    Maximum_Salary=("Salary", "max")
)

summary = summary.sort_values(
    "Average_Salary",
    ascending=False
)
```

```python
summary
```

### 🟢 Task 7 — Concat Practice

```python
new_employees = pd.DataFrame({
    "ID": [107, 108],
    "Name": ["Grace", "Henry"],
    "Department": ["Data", "AI"],
    "Age": [27, 31]
})

employees1 = pd.concat(
    [employees, new_employees],
    ignore_index=True
)

employees1
```

### 🟢 Task 8 — Handle the New Batch

```python
new_salary = pd.DataFrame({
    "ID": [107, 108],
    "Salary": [88000, 97000],
    "Experience": [4, 7]
})

new_performance = pd.DataFrame({
    "ID": [107, 108],
    "Performance": [91, 94]
})

employees2 = employees1.merge(
    new_salary,
    on="ID"
)

print(employees1)

employees2 = employees1.merge(
    new_performance,
    on="ID"
)
```

```python
employees2
```

### 🔥 Task 9 — Final Combined Dataset

```python
df = employees.merge(salary, on="ID")
df = df.merge(performance, on="ID")

new_df = new_employees.merge(new_salary, on="ID")
new_df = new_df.merge(new_performance, on="ID")

final_df = pd.concat([df, new_df], ignore_index=True)
```

```python
final_df
```

```python
final_df["Annual_Bonus"] = np.where(
    final_df["Salary"] >= 90000,
    10000,
    5000
)

final_df["Total_Compensation"] = (
    final_df["Salary"] + final_df["Annual_Bonus"]
)
```

```python
final_df["Performance_Level"] = np.where(
    final_df["Performance"] >= 90,
    "Excellent",
    np.where(
        final_df["Performance"] >= 80,
        "Good",
        "Needs Improvement"
    )
)

final_df = final_df.sort_values(
    "Total_Compensation",
    ascending=False
)

final_df
```

### 🔥 Task 10 — Save It

```python
final_df.to_csv("employee_analysis.csv", index=False)
```
